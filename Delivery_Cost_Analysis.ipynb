```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Delivery_Logistics Dataset.csv')
```

```python
df.head()
```

```python
df["delivery_status"].unique()
```

```python
df2 = df[df["delivery_status"] == "failed"]
```

```python
failed_counts = df2.groupby("delivery_partner").size()
```

```python
cost_per_provider = df.groupby('delivery_partner')['delivery_cost'].sum().sort_values(ascending=False)
```

```python
total_shipments = f"{len(df):,}"
```

```python
total_failed = f"{len(df2):,}"
```

```python
total_cost = f"{df['delivery_cost'].sum():,}"
```

```python
print("Number of failed deliveries:")
df2["delivery_partner"].value_counts()
```

```python
print("\nKPI Summary:")
print(f"Total shipments: {total_shipments}")
print(f"Total failed deliveries: {total_failed}")
print(f"Total delivery cost: USD {total_cost}")
```

```python
sns.barplot(x=failed_counts.index,
    y=failed_counts.values,
    color="steelblue")
```

```python
sns.barplot(x=cost_per_provider.index,
    y=cost_per_provider.values,
    color="steelblue")
```
